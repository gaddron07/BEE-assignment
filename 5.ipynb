```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Read the BED file (compressed)
file_name = "mapped.bed"
df = pd.read_csv(file_name, sep="\t", header=None, compression='infer')

# Step 2: Compute C1, C2, X (C2 - C1), and Y (fragment size)
C1 = (df[2] + df[3]) / 2
C2 = (df[8] + df[9]) / 2
X = C2 - C1
Y = df[9] - df[8]

# Optional: Filter unrealistic values (optional cleanup)
valid = (Y > 0) & (Y < 1000) & (np.abs(X) < 2000)
X = X[valid]
Y = Y[valid]

# Step 3: Create a 2D histogram (heatmap data)
x_bins = np.arange(int(X.min()), int(X.max()) + 1, 20)  # bin width of 20
y_bins = np.arange(int(Y.min()), int(Y.max()) + 1, 10)  # bin width of 10

heatmap, xedges, yedges = np.histogram2d(Y, X, bins=[y_bins, x_bins])  # Note: Y is vertical axis

# Step 4: Plot the heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    heatmap,
    xticklabels=np.round(xedges[:-1], 1),
    yticklabels=np.round(yedges[:-1], 1),
    cmap="viridis",
    cbar_kws={'label': 'Fragment Count'}
)
plt.xlabel("X (C2 - C1; Center Distance)")
plt.ylabel("Y (Fragment Size)")
plt.title("Protein Binding V-Plot Heatmap")
plt.tight_layout()
plt.show()
```
