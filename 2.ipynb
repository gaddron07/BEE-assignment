```python
from collections import defaultdict

# Initialize a nested dictionary to count transitions
transitions = defaultdict(lambda: defaultdict(int))

# Valid DNA bases
valid_bases = {'A', 'C', 'G', 'T'}

# Read the FASTA file and build the transition matrix
with open('multiline_input.fasta', 'r') as file:
    prev_base = None
    for line in file:
        if line.startswith('>'):
            continue  # Skip headers
        line = line.strip().upper()
        for base in line:
            if base not in valid_bases:
                prev_base = None  # Reset on invalid characters like N or gaps
                continue
            if prev_base:
                transitions[prev_base][base] += 1
            prev_base = base

# Print transition matrix as probabilities
print("Markov Transition Matrix (Probabilities):\n")
bases = sorted(valid_bases)
print(f"{'From/To':>8}", end='')
for b in bases:
    print(f"{b:>8}", end='')
print()

for from_base in bases:
    total = sum(transitions[from_base].values())
    print(f"{from_base:>8}", end='')
    for to_base in bases:
        prob = transitions[from_base][to_base] / total if total > 0 else 0
        print(f"{prob:8.4f}", end='')
    print()
```
